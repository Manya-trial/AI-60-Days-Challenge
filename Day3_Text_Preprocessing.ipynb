```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from nltk.tokenize import word_tokenize

text = "Machine learning is amazing! It helps computers learn from data, and it powers search engines, chatbots, and recommendations."

print("BEFORE:", text)
tokens = word_tokenize(text)
print("AFTER (tokens):", tokens)
```

```python
import string
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

lower_tokens = [t.lower() for t in tokens]
print("Lowercased:     ", lower_tokens)

no_punct = [t for t in lower_tokens if t not in string.punctuation]
print("No punctuation: ", no_punct)

clean_tokens = [t for t in no_punct if t not in stop_words]
print("No stop words:  ", clean_tokens)
```

```python
from collections import Counter

bow = Counter(clean_tokens)
print("Word counts:", dict(bow))

vocab = sorted(set(clean_tokens))
vector = [clean_tokens.count(w) for w in vocab]
print("Vocabulary:", vocab)
print("Vector:    ", vector)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

docs = [
    "I love learning about AI and machine learning!",
    "Chatbots use language models to answer questions.",
    "Search engines rank pages using text data.",
    "Spam filters classify emails as spam or not spam.",
    "Data cleaning is the first step in every NLP project."
]

def preprocess(text):
    tokens = word_tokenize(text.lower())
    return [t for t in tokens if t not in string.punctuation and t not in stop_words]

df = pd.DataFrame({"original": docs})
df["tokens"] = df["original"].apply(preprocess)
df["cleaned_text"] = df["tokens"].apply(lambda x: " ".join(x))
display(df)

cv = CountVectorizer()
X = cv.fit_transform(df["cleaned_text"])
bow_df = pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out())
display(bow_df)

df.to_csv("preprocessed_text.csv", index=False)
bow_df.to_csv("preprocessed_bow.csv", index=False)
```
